## 6. Referência versus cópia

### Referência

```python
a = [10, 20, 30]
b = a
b.append(40)

print(a)
print(b)
```

Resultado:

```text
[10, 20, 30, 40]
[10, 20, 30, 40]
```

### Cópia

```python
a = [10, 20, 30]
b = a.copy()
b.append(40)

print(a)
print(b)
```

Resultado:

```text
[10, 20, 30]
[10, 20, 30, 40]
```

2° Resolução

````python
class Paciente:
    """Representa um Nó (Node) na lista encadeada."""
    def __init__(self, nome, idade, prioridade):
        self.nome = nome
        self.idade = idade
        self.prioridade = prioridade
        self.proximo = None  # Este é o "ponteiro" para o próximo paciente


class FilaAtendimento:
    """Representa a lista encadeada (Fila) da clínica."""
    def __init__(self):
        self.inicio = None  # Ponteiro para o primeiro paciente da fila
        self.fim = None     # Ponteiro para o último paciente da fila
        self._tamanho = 0   # Contador de elementos

    def ta_vazia(self):
        """Verifica se a fila está vazia."""
        return self.inicio is None

    def adicionar_paciente(self, nome, idade, prioridade):
        """Adiciona um paciente no final da fila."""
        novo_paciente = Paciente(nome, idade, prioridade)

        if self.ta_vazia():
            # Se está vazia, o novo paciente é tanto o início quanto o fim
            self.inicio = novo_paciente
            self.fim = novo_paciente
        else:
            # Aponta o 'proximo' do antigo fim para o novo nó
            self.fim.proximo = novo_paciente
            # Atualiza o ponteiro 'fim' para ser o novo nó
            self.fim = novo_paciente

        self._tamanho += 1

    def atender_paciente(self):
        """Remove e retorna o primeiro paciente da fila."""
        if self.ta_vazia():
            print("A fila está vazia! Nenhum paciente para atender.")
            return None

        # Guardamos a referência do primeiro paciente
        paciente_atendido = self.inicio

        # Movemos o ponteiro 'inicio' para o próximo nó da lista
        self.inicio = self.inicio.proximo

        # Se após atender a fila ficou vazia, ajustamos o ponteiro 'fim'
        if self.inicio is None:
            self.fim = None

        self._tamanho -= 1
        return paciente_atendido

    def listar_pacientes(self):
        """Percorre a lista do início ao fim usando os ponteiros."""
        if self.ta_vazia():
            print("Nenhum paciente na fila.")
            return

        atual = self.inicio  # Ponteiro auxiliar para navegação
        print("\n--- PACIENTES NA FILA ---")
        while atual is not None:
            print(f"{atual.nome} — {atual.idade} anos — {atual.prioridade}")
            atual = atual.proximo  # Avança o ponteiro para o próximo nó
        print("-------------------------\n")

    def tamanho(self):
        """Informa a quantidade de pacientes na fila."""
        return self._tamanho
````

**23. Desafio — Atendimento prioritário**

**1-** Deve modificar o metodo atender_paciente() para que ele insira os pacientes com prioridades antes dos normais e os normais continuam sendo adicionados no final da fila.

**2** Se inicialmente tivermos:

Ana - Normal
Bruno - Normal

A estrutura será:

inicio
  ->
Ana (Normal) → Bruno (Normal) → None

Quando Carlos, que possui prioridade, entrar:

inicio
  ->
Carlos (Prioridade) → Ana (Normal) → Bruno (Normal) → None


**3. Escolha onde o paciente será inserido**
Paciente normal: será inserido no final da fila.
Paciente prioritário: será inserido antes do primeiro paciente normal.
Se a fila estiver vazia, o paciente será o primeiro elemento, independentemente da prioridade.
Se houver vários pacientes prioritários, o novo prioritário será colocado depois dos prioritários que já estão na fila, mantendo a ordem de chegada entre eles.

````python
class Paciente:
    """Representa um paciente na lista encadeada."""

    def __init__(self, nome, idade, prioridade):
        self.nome = nome
        self.idade = idade
        self.prioridade = prioridade
        self.proximo = None


class FilaAtendimento:
    """Representa a lista encadeada (Fila) da clínica."""

    def __init__(self):
        self.inicio = None
        self.fim = None
        self._tamanho = 0

    def ta_vazia(self):
        """Verifica se a fila está vazia."""
        return self.inicio is None

    def adicionar_paciente(self, nome, idade, prioridade):
        """Adiciona um paciente respeitando a prioridade."""

        novo_paciente = Paciente(nome, idade, prioridade)

        # Caso a fila esteja vazia
        if self.ta_vazia():
            self.inicio = novo_paciente
            self.fim = novo_paciente

        # Caso o paciente seja prioritário
        elif prioridade == "Prioridade":

            # Se o primeiro paciente for normal,
            # o novo paciente entra no início
            if self.inicio.prioridade != "Prioridade":
                novo_paciente.proximo = self.inicio
                self.inicio = novo_paciente

            else:
                # Procuramos o último paciente prioritário
                atual = self.inicio

                while (
                    atual.proximo is not None
                    and atual.proximo.prioridade == "Prioridade"
                ):
                    atual = atual.proximo

                # Inserimos o novo paciente depois dos prioritários
                novo_paciente.proximo = atual.proximo
                atual.proximo = novo_paciente

                # Caso tenha sido inserido no final
                if novo_paciente.proximo is None:
                    self.fim = novo_paciente

        # Caso o paciente seja normal
        else:
            # Adiciona no final da fila
            self.fim.proximo = novo_paciente
            self.fim = novo_paciente

        self._tamanho += 1

    def atender_paciente(self):
        """Remove e retorna o primeiro paciente da fila."""

        if self.ta_vazia():
            print("A fila está vazia! Nenhum paciente para atender.")
            return None

        paciente_atendido = self.inicio

        # O segundo paciente passa a ser o primeiro
        self.inicio = self.inicio.proximo

        # Se a fila ficou vazia
        if self.inicio is None:
            self.fim = None

        self._tamanho -= 1

        return paciente_atendido

    def listar_pacientes(self):
        """Exibe todos os pacientes da fila."""

        if self.ta_vazia():
            print("Nenhum paciente na fila.")
            return

        atual = self.inicio

        print("\n--- PACIENTES NA FILA ---")

        while atual is not None:
            print(
                f"{atual.nome} — "
                f"{atual.idade} anos — "
                f"{atual.prioridade}"
            )

            atual = atual.proximo

        print("-------------------------\n")

    def tamanho(self):
        """Retorna a quantidade de pacientes na fila."""
        return self._tamanho


# =========================
# TESTE
# =========================

fila = FilaAtendimento()

fila.adicionar_paciente("Ana", 30, "Normal")
fila.adicionar_paciente("Bruno", 40, "Normal")
fila.adicionar_paciente("Carlos", 25, "Prioridade")

fila.listar_pacientes()

# Atendendo os pacientes
paciente = fila.atender_paciente()

if paciente:
    print(f"Paciente atendido: {paciente.nome}")

fila.listar_pacientes()
````

**O que é uma referência em Python?**
 - É uma associação entre uma variavel para um objeto na memória, podendo a variavel acessar esse objeto

**Qual a diferença entre b = a e b = a.copy()?**
- Um referencia o mesmo endereço de memória e o outra não, criando uma cópia

**Por que uma estrutura encadeada precisa de referências?**
- Pois diferentemente de uma lista sequencial ela precisa de referencia para saber qual é o proximo elemento da estrutura

**Qual erro você encontrou durante a depuração?**
- é necessario passar o valor do proxima referencia para a variavel atual no momento de percorrer uma lista encadeada se não o progama entra em loop infinito

**Como referências ajudam a compreender estruturas dinâmicas?**
- Estruturas dinâmicas utilizam referencia para poder crescer de tamanho, não sendo necessaria estarem juntas na memória
