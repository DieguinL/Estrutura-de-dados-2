Exercício 1

```python
class PilhaEstatica:
    def __init__(self, capacidade: int):
        if capacidade <= 0:
            raise ValueError("A capacidade deve ser maior que zero.")
        self.capacidade = capacidade
        self.array = [None] * capacidade
        self.top = -1  # Ponteiro manual do topo

    def is_empty(self) -> bool:
        return self.top == -1

    def is_full(self) -> bool:
        return self.top == self.capacidade - 1

    def push(self, valor):
        if valor is None:
            raise ValueError("Não é permitido empilhar valores nulos (None).")

        if self.is_full():
            print(f"Erro: Overflow da pilha! Não foi possível inserir {valor}.")
            return False

        self.top += 1
        self.array[self.top] = valor
        print(f"Push: {valor} inserido na posição {self.top}.")
        return True

    def pop(self):
        if self.is_empty():
            print("Erro: Underflow da pilha! A pilha está vazia.")
            return None

        valor = self.array[self.top]
        self.array[self.top] = None  # Limpa a referência
        self.top -= 1
        return valor

    def peek(self):
        if self.is_empty():
            print("A pilha está vazia.")
            return None
        return self.array[self.top]

# Testando a classe corrigida
pilha = PilhaEstatica(capacidade=3)

pilha.push(10)
pilha.push(20)
pilha.push(30)
pilha.push(40)  # Testa o overflow

print(f"Elemento no topo (peek): {pilha.peek()}")
print(f"Elemento desempilhado: {pilha.pop()}")
print(f"Elemento desempilhado: {pilha.pop()}")
print(f"O topo atual é: {pilha.peek()}")
```

EXERCÍCIO 2

```python
from collections import deque

class SpoolerImpressora:
    def __init__(self):
        # Utiliza deque para eficiência de remoção O(1) no início da fila
        self.fila_impressao = deque()

    def adicionar_documento(self, nome_documento: str):
        """Adiciona um documento ao final da fila de impressão."""
        self.fila_impressao.append(nome_documento)
        print(f"Documento '{nome_documento}' foi adicionado à fila de impressão.")

    def imprimir_documento(self):
        """Remove e imprime o documento mais antigo da fila."""
        if not self.fila_impressao:
            print("Nenhum documento pendente para impressão.")
            return None

        doc_impresso = self.fila_impressao.popleft()
        print(f"Imprimindo: '{doc_impresso}'...")
        return doc_impresso

    def mostrar_fila(self):
        """Exibe o estado atual da fila de impressão."""
        if not self.fila_impressao:
            print("Fila de impressão está vazia.")
        else:
            print(f"Fila atual: {list(self.fila_impressao)}")


# --- Teste do Sistema de Impressão ---
spooler = SpoolerImpressora()

# Adicionando documentos à fila
spooler.adicionar_documento("Relatorio_Final.pdf")
spooler.adicionar_documento("Trabalho_Estrutura_Dados.docx")
spooler.adicionar_documento("Comprovante_Inscricao.pdf")

print("\n--- Estado da Fila ---")
spooler.mostrar_fila()

print("\n--- Processando Impressões ---")
spooler.imprimir_documento()  # Imprime: Relatorio_Final.pdf
spooler.imprimir_documento()  # Imprime: Trabalho_Estrutura_Dados.docx

print("\n--- Estado da Fila ---")
spooler.mostrar_fila()
```
